```python
def get_intervention(row):
    if row['risk_level'] == 'High Risk':
        if row['AttendancePercent'] < 60:
            return "Immediate counselor meeting — attendance issue"
        elif row['IsDeclining']:
            return "Academic support — performance declining"
        else:
            return "General academic support needed"
    elif row['risk_level'] == 'Medium Risk':
        return "Monitor closely, encourage engagement"
    else:
        return "On track, no action needed"

df['Intervention'] = df.apply(get_intervention, axis=1)
```

```python
def get_student_profile(student_id):
    student = df[df['StudentID'] == student_id]
    if student.empty:
        return None
    return student.iloc[0]

# Try it
profile = get_student_profile(5)
print(profile)
```

```python
def risk_breakdown(row):
    return {
        'Attendance factor': (100 - row['AttendancePercent']) * 0.3,
        'Assignment factor': (100 - row['AssignmentSubmissionRate']) * 0.25,
        'Score factor': (100 - row['Avg_Score']) * 0.3,
        'Declining penalty': 15 if row['IsDeclining'] else 0
    }

breakdown = risk_breakdown(profile)
print(breakdown)
```

```python
def compare_to_class(row):
    class_avg = df[df['Class'] == row['Class']]['RiskScore'].mean()
    diff = row['RiskScore'] - class_avg
    return diff

print(compare_to_class(profile))
```

```python
def calc_risk(attendance, assignment_rate, avg_score, is_declining):
    return (
        (100 - attendance) * 0.3 +
        (100 - assignment_rate) * 0.25 +
        (100 - avg_score) * 0.3 +
        is_declining.astype(int) * 15
    )
```

```python
attendance_boost = 10
submission_boost = 10

sim_attendance = (df['AttendancePercent'] + attendance_boost).clip(upper=100)
sim_submission = (df['AssignmentSubmissionRate'] + submission_boost).clip(upper=100)

df['RiskScore_After'] = calc_risk(sim_attendance, sim_submission, df['Avg_Score'], df['IsDeclining'])
df['risk_level_After'] = np.where(df['RiskScore_After'] > 60, 'High Risk',
                          np.where(df['RiskScore_After'] > 35, 'Medium Risk', 'Low Risk'))
```

```python
before_high = (df['risk_level'] == 'High Risk').sum()
after_high = (df['risk_level_After'] == 'High Risk').sum()

print(f"High Risk before: {before_high}")
print(f"High Risk after: {after_high}")
print(f"Students moved out of High Risk: {before_high - after_high}")
```

```python
def add_risk_columns(df):
    df["AttendancePercent"] = df["AttendancePercent"].fillna(df["AttendancePercent"].median())
    df["AssignmentSubmissionRate"] = df["AssignmentSubmissionRate"].fillna(df["AssignmentSubmissionRate"].median())

    df["Avg_Score"] = df[["Test1_Score", "Test2_Score", "Test3_Score"]].mean(axis=1)
    df["Trend"] = df["Test3_Score"] - df["Test1_Score"]
    df["IsDeclining"] = df["Trend"] < -10

    df["RiskScore"] = (
        (100 - df["AttendancePercent"]) * 0.3
        + (100 - df["AssignmentSubmissionRate"]) * 0.25
        + (100 - df["Avg_Score"]) * 0.3
        + df["IsDeclining"].astype(int) * 15
    )

    df["risk_level"] = np.where(
        df["RiskScore"] > 60, "High Risk",
        np.where(df["RiskScore"] > 35, "Medium Risk", "Low Risk")
    )
    return df
```
